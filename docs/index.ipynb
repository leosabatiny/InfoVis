Europe’s energy landscape is changing fast. Until now, most of our electricity has come from coal, gas and oil—but rising concern over climate change means renewables like wind, solar and hydro are taking a front seat (Saraji et al., 2023). In this project, we look at how Europe’s energy mix has shifted month by month, how much it costs to build and operate clean-energy systems, and how ready different countries are for the switch. To frame our analysis, we consider two key perspectives:

Climate urgency: The need to accelerate the shift to renewables in order to limit temperature rise and avoid severe consequences on agriculture, ecosystems and extreme weather.

Economic reality: The financial and logistical challenges of replacing existing infrastructure, securing raw materials and funding large-scale renewable projects across diverse economies.

First, we’ll explore trends in monthly electricity generation by source, so you can see exactly how renewables are growing compared to traditional fuels. Next, we’ll break down the investment needed to expand wind farms, solar parks and hydro plants—and compare those figures to national GDPs to understand the financial challenge. Finally, we’ll map which countries have the resources, infrastructure and funding in place, and which countries may struggle as they move away from fossil fuels.

Pulling together data on generation volumes, technology costs and country-level indicators, this analysis will show where Europe stands today and highlight the biggest hurdles on the road to a cleaner, more reliable power system. Along the way, our visuals will bring the numbers to life and help explain what it all means for the continent’s energy future.
