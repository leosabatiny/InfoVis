# Reflection

We focused on improving the clarity of our arguments in the visualizations’ captions, ensuring a balance between interactivity and interpretability, reducing overcrowded visuals by filtering key categories and countries, and restructuring the story to align more closely with reader logic.

We applied design heuristics to evaluate our visual encoding choices, ensuring effective axis scaling, color-blind safe palettes, and proper chart types for each relationship.



## Peer feedback
On June 19th, 2025, our team D4 gathered with 3 other groups for peer feedback, where one group at a time pitched the work that had been done in the past 2 weeks. Our group was the last to give the presentation, which resulted in only one group (D2) giving us peer feedback. After the pitch, group D2 and our teaching assistant gave us feedback, which we use to improve our draft for the data story and finally craft the final version.

Below is a list of the feedback provided.

### What has been done well:
- The chosen visualizations looked great.
    - The interactive visualisations were generally appreciated.
- The chosen topic was interesting.
- Bonus points for developing a webpage

### What needed to be improved:
- The perspectives are too much alike.
- Some visualizations could be improved upon.
    - Choosing better, more intuitive colors for each graph (Hydroelectrict power should be blue, because people associate water with the color blue).
- The data story lacks focus, making it hard for the audience to grasp the main takeaway.

Just before the discussion and peer feedback, we came to the realisation that one of our datasets wasn't trustworthy. Right after we received the feedback, we changed our chosen perspectives and started to search for new datasets that would fit our new perspectives.
