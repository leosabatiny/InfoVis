# Reflection

We focused on improving the clarity of our arguments in the visualizations’ captions, ensuring a balance between interactivity and interpretability, reducing overcrowded visuals by filtering key categories and countries, and restructuring the story to align more closely with reader logic.

We used design rules to choose clear scales, safe colors, and suitable chart types.



## Peer Feedback

On June 19th, 2025, team D4 presented its progress alongside three other groups. Due to time constraints, only group D2 and our teaching assistant gave us feedback, which we used to improve our data story and finalize the project.

### Positive Feedback:
- Visually appealing and interactive graphs  
- Interesting topic  
- Bonus points for developing a website

### Points for Improvement:
- Perspectives were too similar  
- Some graphs needed better, more intuitive color choices (e.g., blue for hydropower)  
- The data story lacked focus and a clear message

Right before the session, we discovered issues with one dataset. After the feedback, we revised our perspectives and selected new datasets to support a clearer narrative.


## Self-reflection
If this project lasted 1-3 weeks longer, we would have shifted our focus from Europe to the world. It would have been interesting to compare countries from different continents. We would probably keep our current visualisations and just improve upon them and add more countries.
