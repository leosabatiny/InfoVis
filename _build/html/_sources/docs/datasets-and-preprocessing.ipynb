# Datasets and preprocessing

## Datasets

To analyze Europe’s energy transition from both climate urgency and economic feasibility perspectives, we selected three datasets that collectively cover electricity production, emissions, and market prices. Sufficient temporal coverage and standardized country labels were required to enable reliable cross-country comparisons.

### Dataset 1: Monthly Electricity Production by Source

-   **Source:** [IEA Monthly Electricity Statistics](https://www.iea.org/data-and-statistics/data-product/monthly-electricity-statistics)
-   **Original File:** `data2.csv`
-   **Period Covered:** 2010–2025
-   **Size:** 150,850 rows, 6 columns
-   **Key Fields (Relevant to Economic Feasibility & Sustainability Progress):**
    -   `Country`
    -   `Product (energy source)`
    -   `Value (GWh)`
    -   `Balance`
    -   `Time`

This dataset enables granular, source-specific monthly energy trend analysis by country, vital for assessing current energy mixes and the progress of the energy transition towards sustainability. It is fundamental for understanding the scope of energy production that needs to be decarbonized.

---

### Dataset 2: Global CO2 and Greenhouse Gas Emissions

-   **Source:** [GitHub – Our World in Data CO2 Data](https://github.com/owid/co2-data/blob/master/owid-co2-data.csv)
-   **Original File:** `owid-co2-data.csv`
-   **Period Covered:** 1750–2023
-   **Size:** 50,191 rows, 79 columns
-   **Key Variables (Relevant to Economic Feasibility & Climate Necessity):**
    -   `country`
    -   `year`
    -   `gdp`
    -   `population`
    -   `co2`
    -   `co2_per_capita`
    -   `temperature_change_from_co2`
    -   `temperature_change_from_ghg`
    -   `total_ghg`
    -   other fuel-specific emissions 

This dataset provides extensive global data on CO2 and greenhouse gas emissions, alongside socioeconomic indicators and temperature change metrics. It is crucial for quantifying the climate necessity for sustainability and understanding the economic context of emission reduction efforts.

---

### Dataset 3: European Wholesale Electricity Price Data

-   **Source:** [Ember – European Wholesale Electricity Price Data](https://ember-energy.org/data/european-wholesale-electricity-price-data/)
-   **Original File:** `european_wholesale_electricity_price_data_monthly.csv`
-   **Period Covered:** 2015-01 to 2025-06
-   **Size:** 3,690 rows, 4 columns
-   **Key Variables (Relevant to Economic Feasibility):**
    -   `Country`
    -   `Date`
    -   `Price (EUR/MWhe)`

This dataset is essential for analyzing the economic feasibility of energy transition by tracking the fluctuating costs of electricity across European markets, providing insight into the financial burdens and potential market impacts of moving towards sustainable energy.

---

### Dataset 4: Global Land Temperatures by Country

-   **Source:** [Berkeley Earth project](http://berkeleyearth.org/), affiliated with Lawrence Berkeley National Laboratory; available via [Kaggle](https://www.kaggle.com/datasets/berkeleyearth/climate-change-earth-surface-temperature-data)
-   **Original File:** `GlobalLandTemperaturesByCountry.csv`
-   **Period Covered:** 1743–2013
-   **Size:** 577,462 rows, 4 columns
-   **Key Fields (Relevant to Climate Necessity & Sustainability Progress):**
    -   `dt` (Date)
    -   `AverageTemperature` (degrees Celsius)
    -   `AverageTemperatureUncertainty` (degrees Celsius)
    -   `Country`

This dataset provides extensive historical land temperature data by country, which is fundamental for analyzing long-term climate trends, assessing the impact of global warming, and correlating temperature changes with other environmental and human-induced factors. It is crucial for understanding the **climate necessity** for sustainability efforts.

## Cleaning

### Phase 1: Structural Harmonization

We resolved inconsistencies in column names and formats across datasets. Country names were standardized manually (e.g., "Czechia" vs "Czech Republic") to ensure consistent matching. Time fields were unified—converting varied formats like text-based months, year integers, and datetime objects into a common structure.

### Phase 2: Data Quality and Filtering

We filtered all datasets to include only 31 European countries (EU-27 plus Norway, Switzerland, Iceland, and the UK). Aggregate categories were removed to focus on specific energy sources. Missing values were handled contextually—zero-filled for CO₂ subcategories, preserved as NaN for economic indicators. We restricted analysis to years with reliable coverage: emissions from 1950 onward, and electricity prices from 2015.

## Aggregations

### Renewable Share

$\text{renewable\_share} = \frac{\sum(\text{solar, wind, hydro, other renewables})}{\text{total production}}$

### CO₂ Intensity

$\text{co2\_intensity} = \frac{\text{total CO₂ emissions}}{\text{GDP}}$

### Price Volatility

* Monthly rolling standard deviation of prices
* Year-over-year price changes

## Variable Classifications

* **Continuous:** production (GWh), GDP, population, CO₂ emissions, CO₂ per capita, price (EUR/MWhe), renewable_share, co2_intensity
* **Categorical:** Country, Product, Balance
* **Temporal:** year, Date, Time
* **Derived:** energy_intensity, renewable_growth_rate, price_volatility, investment_requirement_pct_gdp
